```python

# Check dataset information
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nStatistical Summary:")
display(df.describe())

print("\nProduct Categories:")
print(df["Product Category"].value_counts())
```

```python

# Step 3: Data Cleaning and Sales Analysis

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Convert Date column to datetime format
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

# Check date conversion
print("Invalid Dates:", df["Date"].isna().sum())

# Check total sales and transactions
print("Total Transactions:", len(df))
print("Total Revenue:", df["Total Amount"].sum())
print("Average Transaction Amount:", round(df["Total Amount"].mean(), 2))
print("Average Customer Age:", round(df["Age"].mean(), 2))

# Sales by product category
category_sales = df.groupby("Product Category")["Total Amount"].sum().sort_values(ascending=False)

print("\nSales by Product Category:")
display(category_sales)

# Transactions by gender
print("\nTransactions by Gender:")
display(df["Gender"].value_counts())

# Monthly sales
df["Month"] = df["Date"].dt.to_period("M").astype(str)
monthly_sales = df.groupby("Month")["Total Amount"].sum().sort_index()

print("\nMonthly Sales:")
display(monthly_sales)
```

```python

# Graph 1: Sales by Product Category
plt.figure(figsize=(8, 5))
category_sales.plot(kind="bar")
plt.title("Total Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# Graph 2: Monthly Sales Trend
plt.figure(figsize=(12, 5))
monthly_sales.plot(kind="line", marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Graph 3: Transactions by Gender
plt.figure(figsize=(6, 5))
df["Gender"].value_counts().plot(kind="bar")
plt.title("Transactions by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# Graph 4: Age Distribution
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="Age", bins=10, kde=True)
plt.title("Customer Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python

# Create a sample retail sales dataset

data = {
    "Transaction ID": [1001, 1002, 1003, 1004, 1005, 1006, 1007, 1008, 1009, 1010],
    "Date": [
        "2026-01-05", "2026-01-12", "2026-02-03", "2026-02-18",
        "2026-03-02", "2026-03-15", "2026-04-06", "2026-04-20",
        "2026-05-10", "2026-05-22"
    ],
    "Customer ID": ["C01", "C02", "C03", "C01", "C04", "C05", "C02", "C06", "C03", "C07"],
    "Gender": ["Female", "Male", "Female", "Male", "Female", "Male", "Female", "Male", "Female", "Male"],
    "Age": [25, 32, 28, 40, 35, 22, 31, 45, 29, 38],
    "Product Category": ["Beauty", "Clothing", "Electronics", "Beauty", "Clothing", "Electronics", "Beauty", "Clothing", "Electronics", "Beauty"],
    "Quantity": [2, 1, 1, 3, 2, 1, 2, 1, 2, 1],
    "Price per Unit": [500, 1200, 5000, 500, 1200, 5000, 500, 1200, 5000, 500],
}

df = pd.DataFrame(data)
df["Total Amount"] = df["Quantity"] * df["Price per Unit"]
df["Date"] = pd.to_datetime(df["Date"])

print("Dataset created successfully!")
display(df.head())
```

```python

# Monthly Sales Analysis

df["Month"] = df["Date"].dt.to_period("M")

monthly_sales = df.groupby("Month")["Total Amount"].sum()

print("Monthly Sales:")
display(monthly_sales)

# Create a line chart
plt.figure(figsize=(10, 5))
plt.plot(monthly_sales.index.astype(str), monthly_sales.values, marker="o")

plt.title("Monthly Retail Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales (₹)")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python

# Sales by Product Category

category_sales = df.groupby("Product Category")["Total Amount"].sum()
print("Sales by Product Category:")
display(category_sales)

# Bar chart
plt.figure(figsize=(8, 5))
category_sales.plot(kind="bar")

plt.title("Total Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales (₹)")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

# Customer Demographics Analysis

print("Customer Age Statistics:")
display(df["Age"].describe())

print("\nTransactions by Gender:")
display(df["Gender"].value_counts())

# Age distribution chart
plt.figure(figsize=(8, 5))
plt.hist(df["Age"], bins=5, edgecolor="black")
plt.title("Customer Age Distribution")
plt.xlabel("Customer Age")
plt.ylabel("Number of Transactions")
plt.tight_layout()
plt.show()

# Gender distribution chart
gender_counts = df["Gender"].value_counts()

plt.figure(figsize=(6, 5))
gender_counts.plot(kind="bar")
plt.title("Transactions by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

# Add product names to the sample dataset

product_names = [
    "Face Cream", "T-Shirt", "Headphones", "Face Wash",
    "Jeans", "Smart Watch", "Lipstick", "Jacket",
    "Bluetooth Speaker", "Moisturizer"
]

df["Product Name"] = product_names

# Top 10 products by sales revenue
top_products = (
    df.groupby("Product Name")["Total Amount"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print("Top 10 Products by Sales Revenue:")
display(top_products)

# Bar chart
plt.figure(figsize=(10, 6))
top_products.sort_values().plot(kind="barh")
plt.title("Top 10 Products by Sales Revenue")
plt.xlabel("Total Sales (₹)")
plt.ylabel("Product Name")
plt.tight_layout()
plt.show()
```

```python

# Correlation Heatmap

numeric_columns = [
    "Age",
    "Quantity",
    "Price per Unit",
    "Total Amount"
]

correlation = df[numeric_columns].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Retail Sales Data")
plt.tight_layout()
plt.show()
```
